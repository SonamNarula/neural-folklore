```python
# Dictionary stores data as:
# key : value

student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

print(student)
print(type(student))
```

```python
# Empty dictionary

d = {}

print(d)
print(type(d))
```

```python
student = {
    "name": "Sonam",
    "age": 21,
    "cgpa": 8.5
}

print(student)
```

```python
data = {
    "name": "Sonam",     # string
    "age": 21,           # integer
    "cgpa": 8.5,         # float
    "active": True       # boolean
}

print(data)
```

```python
student = {
    "name": "Sonam",
    "age": 21,
    "cgpa": 8.5
}
# "name" → key
# "Sonam" → value

# "age" → key
# 21 → value

# "cgpa" → key
# 8.5 → value
```

```python
d = {
    "name": "Sonam",
    1: "one",
    2.5: "two point five",
    (1, 2): "coordinate"
}

print(d)
```

```python
# Accessing Dictionary Elements

student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

print(student["name"])
print(student["age"])
print(student["branch"])
```
