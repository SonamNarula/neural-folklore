```python
# Accessing Dictionary Elements

student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

print(student["name"])
print(student["age"])
print(student["branch"])
```

```python
# Using get() 

# get() is a safer way to access a dictionary value.

student = {
    "name": "Sonam",
    "age": 21
}

print(student.get("name"))
print(student.get("age"))
print(student.get("cgpa"))
```

```python
# get() with default value
student = {
    "name": "Sonam",
    "age": 21
}

print(student.get("cgpa", 0))
```

```python
# Adding Dictionary Elements

# Simply assign a value to a new key.

student = {
    "name": "Sonam",
    "age": 21
}

student["branch"] = "CSE"

print(student)
```

```python
# Modifying Dictionary Elements

# If the key already exists, assignment updates its value.
student = {
    "name": "Sonam",
    "age": 21
}

student["age"] = 22

print(student)
```

```python
# Deleting Dictionary Elements

# del
student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

del student["age"]

print(student)
```

```python
# pop()

# pop() removes the key and returns its value.

student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

age = student.pop("age")

print("Removed value:", age)
print(student)
```

```python
#  pop() with Default Value

# If the key doesn’t exist:
student = {
    "name": "Sonam"
}

print(student.pop("age", None))
```

```python
#  popitem()

# Removes and returns the last inserted key-value pair.

student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

item = student.popitem()

print("Removed:", item)
print(student)
```

```python
# clear()

# Removes everything.

student = {
    "name": "Sonam",
    "age": 21
}

student.clear()

print(student)
```

```python
# Dictionary Methods

# keys()

# Returns dictionary keys.

student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

print(student.keys())
```

```python
# values()

# Returns all values.

student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

print(student.values())
```

```python
# items() 

# Returns key-value pairs.

student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

print(student.items())
```

```python
# Iterating Over Dictionaries

# Only keys
student = {
    "name": "Sonam",
    "age": 21,
    "branch": "CSE"
}

for key in student:
    print(key)
```

```python
# Keys + values
for key, value in student.items():
    print(key, value)
```

```python
# Only values
for value in student.values():
    print(value)
```

```python
# Dictionary Comprehension 

# Similar to list comprehension.

# Normal approach

squares = {}

for i in range(1, 6):
    squares[i] = i * i

print(squares)
```

```python
# Dictionary comprehension
squares = {i: i * i for i in range(1, 6)}

print(squares)
```

```python
# Dictionary Comprehension with Condition
even_squares = {
    i: i * i
    for i in range(1, 11)
    if i % 2 == 0
}

print(even_squares)
```

```python
# Frequency Map — MOST IMPORTANT FOR DSA 

# This is one of the biggest reasons you need dictionaries.

arr = [1, 2, 2, 3, 3, 3, 4]

freq = {}

for x in arr:

    if x in freq:
        freq[x] += 1
    else:
        freq[x] = 1

print(freq)
```

```python
# Frequency Map Using get()
arr = [1, 2, 2, 3, 3, 3, 4]

freq = {}

for x in arr:
    freq[x] = freq.get(x, 0) + 1

print(freq)
```

```python
# Frequency Map Using setdefault()
arr = [1, 2, 2, 3, 3, 3]

freq = {}

for x in arr:
    freq[x] = freq.setdefault(x, 0) + 1

print(freq)
```

```python
# defaultdict — Very Important Later

# Python also has collections.defaultdict.

from collections import defaultdict

freq = defaultdict(int)

arr = [1, 2, 2, 3, 3, 3]

for x in arr:
    freq[x] += 1

print(freq)
```

```python
# Grouping Values Using Dictionary
words = ["apple", "ant", "banana", "ball", "cat"]

groups = {}

for word in words:

    first_letter = word[0]

    if first_letter not in groups:
        groups[first_letter] = []

    groups[first_letter].append(word)

print(groups)
```
