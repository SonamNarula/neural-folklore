```python
# A tuple is an ordered and immutable collection of elements.

numbers = (10, 20, 30, 40)

print(numbers)
print(type(numbers))

# Accessing elements using index
# Just like C++ arrays/vectors, indexing starts from 0.

print(numbers[0])   # 10
print(numbers[2])   # 30
```

```python
# Normal tuple

t1 = (10, 20, 30)

print(t1)
```

```python
# Parentheses are optional in many cases

t2 = 10, 20, 30

print(t2)
print(type(t2))
```

```python
# Empty tuple

empty = ()

print(empty)
print(type(empty))
```

```python
# Correct
t1 = (10,)

print(t1)
print(type(t1))
```

```python
# This is NOT a tuple

t2 = (10)

print(t2)
print(type(t2))
```

```python
(10,)   # tuple
(10)    # integer
```

```python
# Python tuples can store different data types.

student = ("Sonam", 21, 8.5, True)

print(student)

print(student[0])   # String
print(student[1])   # Integer
print(student[2])   # Float
print(student[3])   # Boolean
```

```python
t = ("A", "B", "C", "D", "E")

# Positive indexing

print(t[0])     # A
print(t[1])     # B
print(t[4])     # E
```

```python
t = ("A", "B", "C", "D", "E")
# negative indexing

print(t[-1])    # E
print(t[-2])    # D
print(t[-3])    # C
```

```python
# tuple slicing
t = (10, 20, 30, 40, 50)

print(t[1:4])
```

```python
t = (10, 20, 30, 40, 50)

# First 3 elements
print(t[:3])

# From index 2 till end
print(t[2:])

# Entire tuple
print(t[:])

# Every second element
print(t[::2])

# Reverse tuple
print(t[::-1])
```

```python
# length of a tuple
t = (10, 20, 30, 40)

print(len(t))
```

```python
# concatenation
a = (1, 2, 3)
b = (4, 5, 6)

c = a + b

print(c)
```

```python
# repitition
t = (1, 2, 3)

print(t * 3)
```

```python
# membership

t = (10, 20, 30, 40)

print(20 in t)
print(100 in t)

print(100 not in t)
```

```python
# tumple iteration 

# normally
t = (10, 20, 30, 40)

for x in t:
    print(x)


# index based loop
t = (10, 20, 30, 40)

for i in range(len(t)):
    print(i, t[i])
```

```python
# List → Mutable

arr = [10, 20, 30]

arr[0] = 100

print(arr)

# Tuple → Immutable

t = (10, 20, 30)

# t[0] = 100    # ERROR
```

```python
# Tuple Methods

# Tuples have only two major methods.

# count()

# Counts occurrences of an element.

t = (10, 20, 20, 30, 20, 40)

print(t.count(20))
```

```python
# index()

# Returns the index of the first occurrence.

t = (10, 20, 30, 20, 40)

print(t.index(20))
```

```python
# Built-in Functions with Tuples

t = (5, 2, 9, 1, 7)

print(len(t))
print(min(t))
print(max(t))
print(sum(t))
```

```python
# sorted()
t = (5, 2, 9, 1)

result = sorted(t)

print(result)
print(type(result))
```

```python
# sorted() returns a list, not a tuple.

# If you want a tuple:

t = (5, 2, 9, 1)

result = tuple(sorted(t))

print(result)
print(type(result))
```

```python
#  Packing

# Packing means putting multiple values into a tuple.

student = "Sonam", 21, "CSE"

print(student)
print(type(student))

# Python automatically creates:("Sonam", 21, "CSE")
```

```python
#  Unpacking

# Take values from a tuple and assign them to variables.

student = ("Sonam", 21, "CSE")

name, age, branch = student

print(name)
print(age)
print(branch)
```

```python
#  Unpacking Must Match

t = (10, 20, 30)

a, b, c = t

print(a)
print(b)
print(c)
```

```python
t = (10, 20, 30)

a, b = t
```

```python
a, b, c, d = t
```

```python
# Unpacking 
# This is very useful.

t = (10, 20, 30, 40, 50)

first, *middle, last = t

print(first)
print(middle)
print(last)
```

```python
# swapping variables
a = 10
b = 20

a, b = b, a

print(a)
print(b)
```

```python
#  Function Returning Multiple Values 

# Very important for Python DSA/AI code.

def get_min_max(numbers):

    minimum = min(numbers)
    maximum = max(numbers)

    return minimum, maximum


result = get_min_max([10, 5, 30, 2, 20])

print(result)
```

```python
def get_min_max(numbers):

    return min(numbers), max(numbers)


minimum, maximum = get_min_max([10, 5, 30, 2, 20])

print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
# Nested Tuples

# A tuple can contain other tuples.

matrix = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
)

print(matrix)
```
