```python
# A tuple is an ordered and immutable collection of elements.

numbers = (10, 20, 30, 40)

print(numbers)
print(type(numbers))

# Accessing elements using index
# Just like C++ arrays/vectors, indexing starts from 0.

print(numbers[0])   # 10
print(numbers[2])   # 30
```

```python
# Normal tuple

t1 = (10, 20, 30)

print(t1)
```

```python
# Parentheses are optional in many cases

t2 = 10, 20, 30

print(t2)
print(type(t2))
```

```python
# Empty tuple

empty = ()

print(empty)
print(type(empty))
```

```python
# Correct
t1 = (10,)

print(t1)
print(type(t1))
```

```python
# This is NOT a tuple

t2 = (10)

print(t2)
print(type(t2))
```

```python
(10,)   # tuple
(10)    # integer
```
