```python
# ============================================
# 1. INTRODUCTION TO LISTS
# ============================================

# List is an ordered and mutable collection of elements.
# Lists are written using square brackets [ ]

numbers = [10, 20, 30, 40, 50]

print(numbers)

# Indexing starts from 0
print(numbers[0])   # 10
print(numbers[1])   # 20
print(numbers[4])   # 50

# Lists can contain duplicate values
marks = [90, 85, 90, 75, 90]

print(marks)

# Lists can contain different data types
mixed = [10, "Python", 3.14, True]

print(mixed)
```

```python
# ============================================
# 2. CREATING LISTS
# ============================================

# Empty list
empty_list = []

print(empty_list)


# List with integers
numbers = [10, 20, 30, 40]

print(numbers)


# List with strings
names = ["Sonam", "Rahul", "Aman"]

print(names)


# List with mixed data types
data = [10, "Python", 3.14, True]

print(data)


# Creating a list using list()
numbers2 = list([1, 2, 3, 4])

print(numbers2)


# Creating a list from a string
# Each character becomes an element

word = list("Python")

print(word)

# Output:
# ['P', 'y', 't', 'h', 'o', 'n']


# Creating a list using range()
numbers3 = list(range(1, 6))

print(numbers3)

# Output:
# [1, 2, 3, 4, 5]
```

```python
# ============================================
# 3. ACCESSING LIST ELEMENTS
# ============================================

numbers = [10, 20, 30, 40, 50]

# Positive indexing
print(numbers[0])  # First element
print(numbers[1])  # Second element
print(numbers[2])  # Third element

# Last element
print(numbers[4])


# Negative indexing
# -1 = last element
# -2 = second last element

print(numbers[-1])  # 50
print(numbers[-2])  # 40
print(numbers[-3])  # 30
```

```python
# example

fruits = ["apple", "banana", "mango", "orange", "grapes"]

print(fruits[0])
print(fruits[2])
print(fruits[-1])
print(fruits[-3])
```

```python
# ============================================
# 4. MODIFYING LIST ELEMENTS
# ============================================

numbers = [10, 20, 30, 40, 50]

print("Before:", numbers)

# Change element at index 2
numbers[2] = 100

print("After:", numbers)

# Original:
# [10, 20, 30, 40, 50]

# After modification:
# [10, 20, 100, 40, 50]
```

```python
# modifying multiple elements

numbers = [10, 20, 30, 40, 50]

numbers[0] = 100
numbers[4] = 500

print(numbers)
```

```python
# hence lists are mutable

# Mutable means we can change the list
# after creating it.

names = ["Aman", "Rahul", "Riya"]

names[1] = "Sonam"

print(names)
```

```python
# ============================================
# 5. LIST METHODS
# ============================================

numbers = [10, 20, 30]

# --------------------------------------------
# append()
# Adds ONE element at the end
# --------------------------------------------

numbers.append(40)

print(numbers)

# [10, 20, 30, 40]
```

```python
numbers = [10, 20, 30]

# Adds multiple elements
numbers.extend([40, 50, 60])

print(numbers)

# [10, 20, 30, 40, 50, 60]
```

```python
# append vs extend

a = [1, 2]

a.append([3, 4])

print(a)

# [1, 2, [3, 4]]


a = [1, 2]

a.extend([3, 4])

print(a)

# [1, 2, 3, 4]
```

```python
# insert
numbers = [10, 20, 40]

# Insert 30 at index 2
numbers.insert(2, 30)

print(numbers)

# [10, 20, 30, 40]

# remove
numbers = [10, 20, 30, 20]

# Removes the FIRST occurrence of 20
numbers.remove(20)

print(numbers)

# [10, 30, 20]


# pop
numbers = [10, 20, 30, 40]

# Removes element at index 1
removed = numbers.pop(1)

print("Removed:", removed)
print("List:", numbers)
```

```python
# clear
numbers = [10, 20, 30]

numbers.clear()

print(numbers)
# []

# index
numbers = [10, 20, 30, 40]

position = numbers.index(30)

print(position)

# 2


# count
numbers = [10, 20, 20, 30, 20]

count = numbers.count(20)

print(count)

# 3



# sort
numbers = [50, 10, 40, 20, 30]

numbers.sort()

print(numbers)

# [10, 20, 30, 40, 50]

# sort descending
numbers = [50, 10, 40, 20, 30]

numbers.sort(reverse=True)

print(numbers)

# [50, 40, 30, 20, 10]



# reverse
numbers = [1, 2, 3, 4, 5]

numbers.reverse()

print(numbers)

# [5, 4, 3, 2, 1]
```

```python
# ============================================
# 6. SLICING LISTS
# ============================================

numbers = [10, 20, 30, 40, 50]

# Syntax:
# list[start : stop : step]

# --------------------------------------------
# Basic slicing
# --------------------------------------------

print(numbers[1:4])

# [20, 30, 40]

# IMPORTANT:
# start is included
# stop is EXCLUDED
```
