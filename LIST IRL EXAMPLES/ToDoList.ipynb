```python
# To-Do List

to_do_list = ["buy groceries", "clean the house", "pay bills"]

# adding a task
to_do_list.append("schedule a meeting")
to_do_list.append("go for a run")

print(to_do_list)


# removing a completed task
to_do_list.remove("clean the house")

print(to_do_list)


# checking if a task is in the list
if "pay bills" in to_do_list:
    print("Pay bills is still pending")


# accessing a particular task
print("First task:", to_do_list[0])


# changing a task
to_do_list[0] = "buy vegetables"

print(to_do_list)


# checking number of tasks
print("Total tasks:", len(to_do_list))
```
