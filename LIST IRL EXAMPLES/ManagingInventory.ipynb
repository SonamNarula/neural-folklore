```python
# Store inventory items in a list
inventory = ["Laptop", "Mouse", "Keyboard", "Monitor"]

print("Current Inventory:")
print(inventory)

# Add a new item
inventory.append("Headphones")

print("\nAfter adding Headphones:")
print(inventory)

# Remove an item
inventory.remove("Mouse")

print("\nAfter removing Mouse:")
print(inventory)

# Check if an item is available
if "Laptop" in inventory:
    print("\nLaptop is available in the inventory.")
else:
    print("\nLaptop is not available.")
```
