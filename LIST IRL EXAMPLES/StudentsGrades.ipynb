```python
# Example 2: Organizing Student Grades

# storing student grades in a list
grades = [85, 72, 90, 68, 95]

print("Student Grades:", grades)


# accessing grades using index
print("First grade:", grades[0])
print("Last grade:", grades[-1])


# accessing multiple grades using slicing
print("First three grades:", grades[:3])
print("Last two grades:", grades[-2:])


# changing a grade
grades[1] = 75

print("\nAfter changing the second grade:", grades)


# adding a new grade at the end
grades.append(88)

print("After adding a new grade:", grades)


# adding a grade at a particular position
grades.insert(2, 91)

print("After inserting a new grade:", grades)


# checking if a grade is present
if 95 in grades:
    print("95 is present in the list")


# removing a particular grade
grades.remove(68)

print("After removing 68:", grades)


# removing a grade using its index
removed_grade = grades.pop(0)

print("Removed grade:", removed_grade)
print("After pop:", grades)


# finding the number of grades
print("\nNumber of grades:", len(grades))


# calculating total marks
total = sum(grades)

print("Total marks:", total)


# calculating average marks
average = total / len(grades)

print("Average marks:", average)


# finding highest and lowest grade
print("Highest grade:", max(grades))
print("Lowest grade:", min(grades))


# sorting grades
grades.sort()

print("\nGrades in ascending order:", grades)


# reversing the list
grades.reverse()

print("Grades in reverse order:", grades)


# printing each grade using a loop
print("\nAll grades:")

for grade in grades:
    print(grade)


# counting students who scored more than 80
count = 0

for grade in grades:
    if grade > 80:
        count += 1

print("\nStudents who scored more than 80:", count)
```
