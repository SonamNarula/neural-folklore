```python
# Store user feedback in a list
feedback = [
    "The app is easy to use",
    "The design is good",
    "The app is slow",
    "I like the features"
]

print("User Feedback:")
print(feedback)

# Add new feedback
feedback.append("The app is very useful")

print("\nUpdated Feedback:")
print(feedback)

# Display each feedback
print("\nAll Feedback:")
for item in feedback:
    print("-", item)

# Check total number of feedbacks
print("\nTotal Feedbacks:", len(feedback))
```
