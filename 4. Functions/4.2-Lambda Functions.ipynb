## 1. What is a Lambda Function?

A lambda function is a small anonymous function.

It is generally used when we need a simple function for a short operation.

Basic syntax:

lambda arguments: expression

```python
# A simple lambda function

square = lambda x: x * x

# Calling the lambda function
print(square(5))
```

### What's done?

- Created a lambda function using the `lambda` keyword.
- `x` is the argument.
- `x * x` is the expression.
- The result is automatically returned.

## 2. Normal Function vs Lambda Function

The same operation can be performed using a normal function and a lambda function.

```python
# Normal function

def square_normal(x):
    return x * x


# Lambda function

square_lambda = lambda x: x * x


print("Normal function:", square_normal(5))
print("Lambda function:", square_lambda(5))
```
