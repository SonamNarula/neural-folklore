## 1. What is a Lambda Function?

A lambda function is a small anonymous function.

It is generally used when we need a simple function for a short operation.

Basic syntax:

lambda arguments: expression

```python
# A simple lambda function

square = lambda x: x * x

# Calling the lambda function
print(square(5))
```

### What's done?

- Created a lambda function using the `lambda` keyword.
- `x` is the argument.
- `x * x` is the expression.
- The result is automatically returned.

## 2. Normal Function vs Lambda Function

The same operation can be performed using a normal function and a lambda function.

```python
# Normal function

def square_normal(x):
    return x * x


# Lambda function

square_lambda = lambda x: x * x


print("Normal function:", square_normal(5))
print("Lambda function:", square_lambda(5))
```

# Normal function

def square_normal(x):
    return x * x


# Lambda function

square_lambda = lambda x: x * x


print("Normal function:", square_normal(5))
print("Lambda function:", square_lambda(5))

### What's done?

Both functions perform the same operation.

Normal function:
- Uses `def`
- Can contain multiple statements

Lambda function:
- Uses `lambda`
- Usually contains one expression
- Automatically returns the result

```python
# Lambda function with one argument

double = lambda number: number * 2

print(double(10))
print(double(25))
```

## 4. Lambda Function with Multiple Arguments

A lambda function can take multiple arguments.

```python
# Lambda function with two arguments

add = lambda a, b: a + b

print(add(10, 20))
```

```python
# Lambda function for multiplication

multiply = lambda a, b: a * b

print(multiply(5, 6))
```

```python
# Lambda function with three arguments

calculate_total = lambda price, quantity, discount: (price * quantity) - discount

price = 500
quantity = 2
discount = 100

total = calculate_total(price, quantity, discount)

print("Total:", total)
```

## 6. Lambda Function with a Condition

We can use a conditional expression inside a lambda function.

Syntax:

lambda x: value_if_true if condition else value_if_false

```python
# Check whether a number is even or odd

check_number = lambda number: "Even" if number % 2 == 0 else "Odd"

print(check_number(10))
print(check_number(7))
```

```python
# Find the larger of two numbers

find_larger = lambda a, b: a if a > b else b

print(find_larger(25, 40))
```

```python
# Find the smaller of two numbers

find_smaller = lambda a, b: a if a < b else b

print(find_smaller(25, 40))
```

```python
# Find the length of a string

find_length = lambda text: len(text)

print(find_length("Python"))
```

```python
# Convert a string to uppercase

make_upper = lambda text: text.upper()

print(make_upper("python"))
```

```python
# Find the square of a number from a list

numbers = [1, 2, 3, 4, 5]

square = lambda x: x * x

for number in numbers:
    print(square(number))
```

## 11. Lambda with map()

`map()` applies a function to every element of an iterable.

Lambda functions are commonly used with `map()`.

Syntax:

map(function, iterable)

```python
# Square every number in the list

numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))

print("Original numbers:", numbers)
print("Squares:", squares)
```

### What's happening?

For every number:

1 → 1 × 1  
2 → 2 × 2  
3 → 3 × 3  
4 → 4 × 4  
5 → 5 × 5

```python
# Convert every name to uppercase

names = ["sonam", "arpit", "rahul", "aman"]

upper_names = list(map(lambda name: name.upper(), names))

print(upper_names)
```

```python
# Add 10 to every number

numbers = [1, 2, 3, 4, 5]

new_numbers = list(map(lambda x: x + 10, numbers))

print(new_numbers)
```

## 14. Lambda with filter()

`filter()` is used to select elements based on a condition.

Syntax:

filter(function, iterable)

The lambda function should return `True` or `False`.


```python
# Get only even numbers

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

```python
# Get only odd numbers

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))

print(odd_numbers)
```

```python
# Get numbers greater than 10

numbers = [5, 12, 8, 20, 3, 15]

large_numbers = list(filter(lambda x: x > 10, numbers))

print(large_numbers)
```

```python
# Get names having more than 5 characters

names = ["Aman", "Sonam", "Rahul", "Alexander", "Priya"]

long_names = list(filter(lambda name: len(name) > 5, names))

print(long_names)
```

## map() vs filter()

`map()` → transforms every element.

`filter()` → selects elements based on a condition.

Example:

map:
[1, 2, 3] → [2, 4, 6]

filter:
[1, 2, 3, 4] → [2, 4]

## 19. Lambda with reduce()

`reduce()` repeatedly applies a function and reduces all values to one final result.

We need to import `reduce` from the `functools` module.

```python
from functools import reduce

# Find the sum of all numbers

numbers = [1, 2, 3, 4, 5]

total = reduce(lambda a, b: a + b, numbers)

print("Total:", total)
```

```python
from functools import reduce

# Find the product of all numbers

numbers = [1, 2, 3, 4, 5]

product = reduce(lambda a, b: a * b, numbers)

print("Product:", product)
```

### How reduce() works

For `[1, 2, 3, 4]`:

First: 1 × 2 = 2  
Then: 2 × 3 = 6  
Then: 6 × 4 = 24

Final answer = 24

## 21. Lambda with sorted()

Lambda functions can be used with `sorted()` to define how we want to sort the data.

```python
# Sort numbers based on their absolute value

numbers = [-10, 5, -2, 8, -1]

sorted_numbers = sorted(numbers, key=lambda x: abs(x))

print(sorted_numbers)
```

```python
# Sort names according to their length

names = ["Sonam", "Aman", "Alexander", "Raj"]

sorted_names = sorted(names, key=lambda name: len(name))

print(sorted_names)
```

```python
# Sort numbers from largest to smallest

numbers = [10, 5, 20, 8, 15]

sorted_numbers = sorted(numbers, key=lambda x: x, reverse=True)

print(sorted_numbers)
```

## 24. Lambda with a List of Tuples

Suppose every tuple contains:

(name, marks)

We can sort students based on their marks.

```python
# Student data

students = [
    ("Sonam", 85),
    ("Aman", 92),
    ("Rahul", 78),
    ("Priya", 95)
]

# Sort students according to marks

sorted_students = sorted(students, key=lambda student: student[1])

print(sorted_students)
```

```python
# Sort students from highest marks to lowest marks

sorted_students = sorted(
    students,
    key=lambda student: student[1],
    reverse=True
)

print(sorted_students)
```

## 26. Lambda with Dictionaries

Lambda functions can also be used to sort a list of dictionaries.

```python
# Student data

students = [
    {"name": "Sonam", "marks": 85},
    {"name": "Aman", "marks": 92},
    {"name": "Rahul", "marks": 78},
    {"name": "Priya", "marks": 95}
]

# Sort students according to their marks

students.sort(key=lambda student: student["marks"])

print(students)
```

## 27. Calling a Lambda Function Directly

We don't always have to store a lambda function in a variable.

We can create and call it directly.

```python
# Lambda function created and called directly

result = (lambda x: x * 2)(10)

print(result)
```
