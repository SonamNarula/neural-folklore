```python
# Function to calculate the total cost of items

def calculate_total(prices):
    total = 0
    
    # Add each item price to the total
    for price in prices:
        total = total + price
    
    return total


# Prices of items in the shopping cart
prices = [250, 120, 80, 500]

# Calling the function
total_cost = calculate_total(prices)

print("Total cost:", total_cost)
```
