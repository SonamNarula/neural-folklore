```python
# Creating a sample text file

with open("sample.txt", "w") as file:
    file.write("Python is easy to learn.\n")
    file.write("Python is powerful.\n")
    file.write("Python is easy and powerful.")


# Function to count the frequency of each word

def count_word_frequency(file_path):
    
    # Dictionary to store the word and its count
    word_count = {}
    
    # Open the file in read mode
    with open(file_path, "r") as file:
        
        # Read the file line by line
        for line in file:
            
            # Split the line into individual words
            words = line.split()
            
            # Go through each word
            for word in words:
                
                # Convert the word to lowercase
                # and remove common punctuation marks
                word = word.lower().strip(".,!?;:\"'")
                
                # Increase the count of the word by 1
                word_count[word] = word_count.get(word, 0) + 1
    
    # Return the final dictionary
    return word_count


# Name of the text file
file_path = "sample.txt"

# Calling the function
word_frequency = count_word_frequency(file_path)

# Display the word frequency
print("Word Frequency:")
print(word_frequency)
```
