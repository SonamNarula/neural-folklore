```python
# variable length arguments
# positional and keywords arguments

def print_numbers(*sonam):
    for number in sonam:
        print(number)
```

```python
print_numbers(1, 2, 3, 4, 5, 6, 7, 8, "sonam")
```

```python
# positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1, 2, 3, 4, 5, 6, 7, 8, "sonam")
```

```python
# keywords arguments

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_details(name = "sonam" , age = "21", country = "germany")
```

```python
def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional argument : {val}")

    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,name = "sonam" , age = "21", country = "germany")
```
