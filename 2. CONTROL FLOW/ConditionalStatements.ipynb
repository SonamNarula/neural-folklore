```python
# if statement
age = 20

if age >= 18:
    print("You are eligible to vote.")
```

```python
# else statement
age = 16

if age >= 18:
    print("You can vote.")
else:
    print("You cannot vote.")
    
```

```python
# elif statement
marks = 75

if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Grade D")
```

```python
# structure

if condition1:
    # code

elif condition2:
    # code

elif condition3:
    # code

else:
    # code
```

```python
# nested conditional statements

age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed.")
    else:
        print("ID required.")
else:
    print("Entry not allowed.")
```

```python
# example
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("Login successful.")
    else:
        print("Wrong password.")
else:
    print("Invalid username.")
```
