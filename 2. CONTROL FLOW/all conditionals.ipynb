```python

age = int(input("Enter your age: "))
marks = int(input("Enter your marks: "))

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")

if marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
else:
    print("Grade: F")

if age >= 18:
    if marks >= 40:
        print("Eligible for admission.")
    else:
        print("Age eligible, but marks are insufficient.")
else:
    print("Not eligible due to age.")
```
