```python
# Python Arithmetic Operators - All in One

a = 10
b = 3

print("a =", a)
print("b =", b)

# Addition
print("Addition (a + b):", a + b)

# Subtraction
print("Subtraction (a - b):", a - b)

# Multiplication
print("Multiplication (a * b):", a * b)

# Division
print("Division (a / b):", a / b)

# Floor Division
print("Floor Division (a // b):", a // b)

# Modulus / Remainder
print("Modulus (a % b):", a % b)

# Exponent / Power
print("Exponent (a ** b):", a ** b)

# Unary Plus
print("Unary Plus (+a):", +a)

# Unary Minus
print("Unary Minus (-a):", -a)
```

```python
# Python Comparison Operators - All in One

a = 10
b = 3

print("a =", a)
print("b =", b)

# Equal to
print("Equal to (a == b):", a == b)

# Not equal to
print("Not equal to (a != b):", a != b)

# Greater than
print("Greater than (a > b):", a > b)

# Less than
print("Less than (a < b):", a < b)

# Greater than or equal to
print("Greater than or equal to (a >= b):", a >= b)

# Less than or equal to
print("Less than or equal to (a <= b):", a <= b)
```

```python
# Python Logical Operators - All in One

a = 10
b = 3

print("a =", a)
print("b =", b)

# AND operator
print("(a > b) and (a == 10):", (a > b) and (a == 10))

# OR operator
print("(a < b) or (a == 10):", (a < b) or (a == 10))

# NOT operator
print("not (a > b):", not (a > b))
```
